```python
#Minimum element in the BinarySearchTree
class BinaryNode:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None
class BinarySearchTree:
  def __init__(self):
    self.root = None
  def insert(self, data):
    if self.root is None:
      self.root = BinaryNode(data)
    else:
      self._insert_recursive(self.root, data)
  def _insert_recursive(self, current_node, data):
    if data < current_node.data:
      if current_node.left is None:
        current_node.left = BinaryNode(data)
      else:
        self._insert_recursive(current_node.left, data)
    elif data > current_node.data:
      if current_node.right is None:
        current_node.right = BinaryNode(data)
      else:
        self._insert_recursive(current_node.right, data)
  def display_inorder(self):
    if self.root is None:
      print("Tree is empty.")
      return
    print("Tree nodes (in-order traversal):")
    self._in_order_display(self.root)
    print("")
  def _in_order_display(self, node):
    if node:
      self._in_order_display(node.left)
      print(node.data, end=" ")
      self._in_order_display(node.right)
  def display_preorder(self):
    if self.root is None:
      print("Tree is empty")
      return
    print("Tree nodes (pre-order traversal):")
    self._pre_order_display(self.root)
    print("")
  def _pre_order_display(self, node):
    if node:
      print(node.data, end=" ")
      self._pre_order_display(node.left)
      self._pre_order_display(node.right)
  def display_postorder(self):
    if self.root is None:
      print("Tree i empty")
      return
    print("Tree nodes (post-order traversa):")
    self._post_order_display(self.root)
  def _post_order_display(self,node):
    if node:
      self._post_order_display(node.left)
      self._post_order_display(node.right)
      print(node.data, end=" ")
  def find_min(self):
    if self.root is None:
      return None
    node = self.root
    while node.left is not None:
      node = node.left
    return node.data
  def find_max(self):
    if self.root is None:
      return None
    node = self.root
    while node.right is not None:
      node = node.right
    return node.data
BST=BinarySearchTree()
BST.insert(5)
BST.insert(3)
BST.insert(7)
BST.insert(2)
BST.insert(4)
BST.insert(6)
BST.insert(8)
BST.display_inorder()
BST.display_preorder()
BST.display_postorder()
min_value=BST.find_min()
max_value=BST.find_max()
print("Maximum value in the BST:", max_value)
print("Minimum value in the BST:", min_value)
```

```python
# Construct a Binary Search Tree from a sorted array
class BinaryNode:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None
def build_bst_from_array(arr):
  if not arr:
    return None
  mid=len(arr)//2
  root=BinaryNode(arr[mid])
  root.left=build_bst_from_array(arr[:mid])
  root.right=build_bst_from_array(arr[mid+1:])
  return root
arr=[12,14,15,17,19,22,26,31,35,42,45,48]
print("Original array:", arr)
BST=build_bst_from_array(arr)
print("In-order traversal of the BST:")
def inorder_traversal(node):
  if node:
    inorder_traversal(node.left)
    print(node.data, end=" ")
    inorder_traversal(node.right)
inorder_traversal(BST)
```

```python
#Graph
class Graph:
  def __init__(self):
    self.graph={}
  def AddVertex(self, vertex):
    if vertex not in self.graph:
      self.graph[vertex]=[]
  def AddEdge(self, vertex1, vertex2):
      self.AddVertex(vertex1)
      self.AddVertex(vertex2)
      self.graph[vertex1].append(vertex2)
      self.graph[vertex2].append(vertex1)
  def ISEdge(self,vertex1, vertex2):
    if vertex1 in self.graph and vertex2 in self.graph[vertex1]:
      return True
    return False
  def RemoveEdge(self, vertex1, vertex2):
    if vertex1 in self.graph and vertex2 in self.graph:
      if vertex2 in self.graph[vertex1]:
        self.graph[vertex1].remove(vertex2)
      if vertex1 in self.graph[vertex2]:
        self.graph[vertex2].remove(vertex1)
    else:
      print(f"Edge between {vertex1} and {vertex2} does not exist or one of the vertices is not in the graph.")
  def display(self):
    for vertex, neighbors in self.graph.items():
      print(f"{vertex}: {neighbors}")
G=Graph()
G.AddEdge('A','B')
G.AddEdge('B','C')
G.AddEdge('C','D')
G.AddEdge('D','E')
G.AddEdge('E','F')
print(G.ISEdge('A','B'))
print(G.ISEdge('B','C'))
print(G.ISEdge('C','D'))
print(G.ISEdge('D','E'))
G.RemoveEdge('E','F')
print("Graph after removing edge E-F:", G.graph)
print("Graph after removing edge A-B:", G.graph)
G.RemoveEdge('Z','Y')
print("Graph after removing edge Z-Y:", G.graph)
G.display()
```
