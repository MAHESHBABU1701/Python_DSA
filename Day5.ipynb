```python
#Sum of Array
arr=[1,2,3,4,5,6]
sum=0
for i in range(len(arr)):
  sum+=arr[i]
print(sum)
```

```python
#Prefix sum
arr=[1,2,3,4,5,5,5,6]
prefix=[]
prefix[0]=arr[0]
prefix[1]=prefix[0]+arr[1]
prefix[2]=prefix[1]+arr[2]
prefix[i]=prefix[i-1]+arr[i]
```

```python
def build_prefix(arr):
  prefix=[arr[0]]
  for i in range(1,len(arr)):
    prefix.append(prefix[i-1]+arr[i])
  return prefix
arr=list(map(int,input().split()))
print(build_prefix(arr))
```

```python
# Sum of range l to r (inclusive, 0-indexed)
def sum_range(prefix,l,r):
  if l==0:
    return prefix[r]
  return prefix[r]-prefix[l-1]
arr=list(map(int,input().split()))
prefix=build_prefix(arr)
print(sum_range(prefix,2,5))
```

```python
# Equillibrium Index
arr1=list(map(int,input().split()))
def equillibrium_index(arr):
  left_sum=0
  total=__builtins__.sum(arr)
  for i in range(len(arr)):
    right_sum=total-arr[i]-left_sum
    if left_sum==right_sum:
      return i
    left_sum+=arr[i]
  return -1
print(equillibrium_index(arr1))
```

```python
#sub_array equals total k
a=list(map(int,input().split()))
target=7
for i in range(len(a) - 1):
  if a[i]+a[i+1]==target:
    print(a[i],a[i+1])
```

```python
#Sub_array equals total k
def subarray(arr, target):
  pairs = []
  n = len(arr)
  for i in range(n):
    for j in range(i + 1, n):
      if arr[i] + arr[j] == target:
        pairs.append((arr[i], arr[j]))
  return pairs
print(f"Original array: {a}")
print(f"Target sum: {target}")
all_pairs = subarray(a, target)
print(f"All pairs that sum to {target}: {all_pairs}")
```
