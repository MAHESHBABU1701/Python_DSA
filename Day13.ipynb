```python
class BinaryNode:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

class BinaryTree:
  def __init__(self):
    self.root = None

  def add(self, data):
    if self.root is None:
      self.root = BinaryNode(data)
      return
    self._recursive_add(self.root, data)

  def _recursive_add(self, current_node, data):
    if current_node.left is None:
      current_node.left = BinaryNode(data)
      return True
    elif current_node.right is None:
      current_node.right = BinaryNode(data)
      return True

    if self._recursive_add(current_node.left, data):
      return True
    elif self._recursive_add(current_node.right, data):
      return True
    return False

  def display_inorder(self):
    if self.root is None:
      print("Tree is empty.")
      return
    print("Tree nodes (in-order traversal):")
    self._in_order_display(self.root)
    print("") # For a new line after display

  def _in_order_display(self, node):
    if node:
      self._in_order_display(node.left)
      print(node.data, end=" ")
      self._in_order_display(node.right)

  def display_preorder(self):
    if self.root is None:
      print("Tree is empty.")
      return
    print("Tree nodes (pre-order traversal):")
    self._pre_order_display(self.root)
    print("")

  def _pre_order_display(self, node):
    if node:
      print(node.data, end=" ")
      self._pre_order_display(node.left)
      self._pre_order_display(node.right)

  def display_postorder(self):
    if self.root is None:
      print("Tree is empty.")
      return
    print("Tree nodes (post-order traversal):")
    self._post_order_display(self.root)
    print("")

  def _post_order_display(self, node):
    if node:
      self._post_order_display(node.left)
      self._post_order_display(node.right)
      print(node.data, end=" ")
my_tree = BinaryTree()
my_tree.add(10)
my_tree.add(5)
my_tree.add(15)
my_tree.add(3)
my_tree.add(7)
my_tree.add(12)
my_tree.add(18)

my_tree.display_inorder()
my_tree.display_preorder()
my_tree.display_postorder()
```
