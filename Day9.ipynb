```python
#Two pointers
#First Occurence of sum of two pairs using two pointers
arr1=[5,6,9,7,4,1,2,3]
arr1.sort()
target=int(input())
left=0
right=len(arr1)-1
found_pair = False
while left<right:
  if arr1[left]+arr1[right]==target:
    print(arr1[left],arr1[right])
    break
    found_pair = True
  elif arr1[left]+arr1[right]<target:
    left+=1
  else:
    right-=1
else:
  print("Pair not found")
```

```python
#Sum of two pairs using two pointers
arr1=[5,6,9,7,4,1,2,3]
arr1.sort()
target=int(input())
left=0
right=len(arr1)-1
found_pair = False
while left<right:
  if arr1[left]+arr1[right]==target:
    print(arr1[left],arr1[right])
    found_pair = True
    left+=1
    right-=1
  elif arr1[left]+arr1[right]<target:
    left+=1
  else:
    right-=1
if not found_pair:
  print("Pair not found")
```

```python
#Find palindrome or not using two pointers
a=input()
left=0
right=len(a)-1
found_palindrome=True
while left<right:
  if a[left]!=a[right]:
    found_palindrome=False
    break
  left+=1
  right-=1
if found_palindrome:
  print("Palindrome")
```
