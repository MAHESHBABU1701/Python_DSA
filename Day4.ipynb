```python
#Count of vowels,consonants,spaces, and special characters
x="Python is a programming language"
count_vowels=0
count_consonants=0
count_spaces=0
count_specialchars=0
for i in range(len(x)):
  if x[i].isalpha():
    if x[i] in "aeiou":
      count_vowels+=1
    else:
      count_consonants+=1
  elif x[i] == " ":
    count_spaces+=1
  else:
    count_specialchars+=1
print(count_vowels)
print(count_consonants)
print(count_spaces)
print(count_specialchars)
```

```python
#Remove Duplicate Strings (order preserved)
n=input("Enter String:")
seen = []
result = ""
for char in n:
    if char not in seen:
        seen.append(char)
        result += char
print(result)
```

```python
#Reverse a string using for loop
n=input("Enter String:")
rev = ""
for char in n:
  rev = char + rev
print(rev)
```

```python
#Anagram
n=input("Enter String:")
w=input("Enter String:")
if sorted(n)==sorted(w):
  print("Anagram")
else:
  print("Not Anagram")
```

```python
#Anagram using dictionary
n=input("Enter String:")
w=input("Enter String:")
dict1={}
dict2={}
for i in n:
  if i in dict1:
    dict1[i]+=1
  else:
    dict1[i]=1
for j in w:
  if j in dict2:
    dict2[j]+=1
  else:
    dict2[j]=1
if dict1==dict2:
  print("Anagram")
else:
  print("Not Anagram")
```

```python
#Recursion
#Factorial
def facorial(n):
  if n<=1:
    return 1
  else:
    return n*facorial(n-1)
n=int(input("Enter Number:"))
print(facorial(n))
```

```python
#Fibonnaci
def fib(n):
  if n<=1:
    return n
  return fib(n-1)+fib(n-2)
n=int(input("Enter Number:"))
for i in range(n):
    print(fib(i),end=" ")
```
