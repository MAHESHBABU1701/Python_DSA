```python
#Count of vowels,consonants,spaces, and special characters
x="Python is a programming language"
count_vowels=0
count_consonants=0
count_spaces=0
count_specialchars=0
for i in range(len(x)):
  if x[i].isalpha():
    if x[i] in "aeiou":
      count_vowels+=1
    else:
      count_consonants+=1
  elif x[i] == " ":
    count_spaces+=1
  else:
    count_specialchars+=1
print(count_vowels)
print(count_consonants)
print(count_spaces)
print(count_specialchars)
```

```python
#Remove Duplicate Strings (order preserved)
n=input("Enter String:")
seen = []
result = ""
for char in n:
    if char not in seen:
        seen.append(char)
        result += char
print(result)
```

```python
#Reverse a string using for loop
n=input("Enter String:")
rev = ""
for char in n:
  rev = char + rev
print(rev)
```

```python
#Anagram
n=input("Enter String:")
w=input("Enter String:")
if sorted(n)==sorted(w):
  print("Anagram")
else:
  print("Not Anagram")
```

```python
#Anagram using dictionary
n=input("Enter String:")
w=input("Enter String:")
dict1={}
dict2={}
for i in n:
  if i in dict1:
    dict1[i]+=1
  else:
    dict1[i]=1
for j in w:
  if j in dict2:
    dict2[j]+=1
  else:
    dict2[j]=1
if dict1==dict2:
  print("Anagram")
else:
  print("Not Anagram")
```

```python
#Recursion
#Factorial
def facorial(n):
  if n<=1:
    return 1
  else:
    return n*facorial(n-1)
n=int(input("Enter Number:"))
print(facorial(n))
```

```python
#Fibonnaci
def fib(n):
  if n<=1:
    return n
  return fib(n-1)+fib(n-2)
n=int(input("Enter Number:"))
for i in range(n):
    print(fib(i),end=" ")
```

```python
#Home work problems
# 1.Find the maximum occurrence character in a string use dictionary
n=input("Enter String:")
dict={}
for i in n:
  if i in dict:
    dict[i]+=1
  else:
    dict[i]=1
max_char=max(dict,key=dict.get)
print(max_char)
```

```python
# 2. sum of digits using recursion
def sum_digits(n):
  if n == 0:
    return 0
  else:
    return (n % 10) + sum_digits(n // 10)
num = int(input("Enter Number: "))
print(sum_digits(num))
```

```python
# 3. compress the string aaabbc->a3b2c1
def compress_string(s):
    compressed = ""
    count = 1

    for i in range(len(s)-1):
      if s[i] == s[i+1]:
          count +=1
      else:
        compressed+=s[i]+str(count)
        count=1
    compressed+=s[-1]+str(count)
    return compressed
print(compress_string("aaabbc"))
```

```python
#Home work
#Pascal triangle
n=int(input())
for i in range(n):
  num=1
  for j in range(n-i-1):
    print("",end=" ")
  for j in range(i+1):
    print(num,end=" ")
    num=num*(i-j)//(j+1)
  print()
```

```python
#Perfect Number
n=int(input("Enter Number:"))
sum=0
for i in range(1,n):
  if n%i==0:
    sum+=i
if sum==n:
  print("Perfect Number")
else:
  print("Not Perfect Number")
```

```python
#Strong Number
n=int(input())
temp=n
s=0
while temp>0:
  digit=temp%10
  fact=1
  for i in range(1,digit+1):
    fact*=i
  s+=fact
  temp//=10
if s==n:
  print("Strong Number")
else:
  print("Not Strong Number")
```
