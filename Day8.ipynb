```python
# Sort the array without using sort() method (Selection Sort)
arr = list(map(int, input().split()))
n = len(arr)
for i in range(n):
    min_idx = i
    for j in range(i + 1, n):
        if arr[min_idx] > arr[j]:
            min_idx = j
    arr[i], arr[min_idx] = arr[min_idx], arr[i]
print(arr)
```

```python
#Merge Sort
def merge_sort(arr):
  if len(arr) > 1:
    mid = len(arr) // 2
    left = arr[:mid]
    right = arr[mid:]
    merge_sort(left)
    merge_sort(right)
    lp=0
    rp=0
    fp=0
    while lp < len(left) and rp < len(right):
      if left[lp] < right[rp]:
        arr[fp] = left[lp]
        lp += 1
        fp += 1
      else:
        arr[fp] = right[rp]
        rp += 1
        fp += 1
    while lp < len(left):
      arr[fp] = left[lp]
      lp += 1
      fp += 1
    while rp < len(right):
      arr[fp] = right[rp]
      rp += 1
      fp += 1
arr=list(map(int,input().split()))
merge_sort(arr)
print(arr)
```

```python
#Quick Sort
import random
def quick_sort(arr):
  if len(arr) <= 1:
    return arr
  pivot = random.choice(arr)
  left = [x for x in arr if x < pivot]
  right = [x for x in arr if x > pivot]
  middle = [x for x in arr if x == pivot]
  return quick_sort(left) + [pivot] + quick_sort(right)
arr=list(map(int,input().split()))
print(quick_sort(arr))
```
