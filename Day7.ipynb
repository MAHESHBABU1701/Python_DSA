```python
arr=list(map(int,input().split()))
target=int(input())
for i in range(len(arr)):
  if target==arr[i]:
    print("Target found at index:",i)
    break
```

```python
#Binary search
a1=list(map(int,input().split()))
target=int(input())
a1.sort()
high=len(a1)-1
low=0
while low<=high:
  mid=(high+low)//2
  if a1[mid]==target:
    print("Target found at index:",mid)
    break
  elif a1[mid]>target:
    high=mid-1
  else:
    low=mid
```

```python
#First Occurence
a2=list(map(int,input().split()))
target=int(input())
a2.sort()
print(a2)
high=len(a2)-1
low=0
result=0
while low<=high:
  mid=(high+low)//2
  if a2[mid]==target:
    result=mid
    high=mid-1
  elif a2[mid]>target:
    high=mid-1
  else:
    low=mid+1
print("Target found at index:",result)
```

```python
#Calculate square root of 50 using binary search
n=50
low = 0
high = n
answer=0
while low<=high:
  mid = (low + high) // 2
  if mid * mid == target:
    answer=mid
    break
  elif mid * mid < n:
    answer=mid
    low = mid+1
  else:
    high = mid-1
print(f"The approximate value of square root {n} is {answer}")
```
